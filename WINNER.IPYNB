```python
def even_sum(numbers:list)-> int:
    total=0
    for i in numbers:
        if i% 2 == 0:
            total+=i
    return total
print(even_sum([3,6,8,1,5,10]))
```

```python
def backward_words(sentences:str)->str :
    words = sentences.split()
    words.reverse()
    return " ".join(words)
print(backward_words('my name is winner,winner is joyful'))

```

```python
def unique_elements(list1: list, list2: list) -> set:
    return set(list1) ^ set(list2)

print(unique_elements([1, 2, 3,4,5,6,7,8,9,0], [2, 4,6,8,]))  
```

```python
def count_vowels(text: str) -> int:
    vowels = "aeiou"
    count = 0
    for char in text.lower():
        if char in vowels:
            count += 1
    return count

print(count_vowels("my name is winner,i am a software developer and data analyst"))  
```

```python
def swap_tuple(t: tuple) -> tuple:
    if len < 2 :
        return t
    lst = list(t)
    lst[0], lst[-1] = lst[-1], lst[0]
    return tuple(lst)

print(swap_tuple((1, 2, 3, 4)))  
```

```python
def is_a_palindrome (words:str) -> bool:
   return words.lower()==words.lower()[::-1]
print(is_a_palindrome("madam"))
print(is_a_palindrome("Level"))
print(is_a_palindrome("winner")) 
```

```python
def sum_of_digits(num: int) -> int:
    return sum(int(digit) for digit in str(num))
print(sum_of_digits(123456789))
```

```python
def names(names: list, min_length: int) -> list:
    return [name for name in names if len(name) >= min_length]

print(names(["Sansa", "Jon snow", "arya stark","brandon","rickon stark", "she is no one"], 4))
```

```python
def merge(dict1:dict, dict2: dict) -> dict:
    result = dict1.copy()
    for key,value in dict2.items():
        result[key] = result.get(key,0) + value
    return result
print(merge({"banana":78, "Mango" : 67,"orange" :90,"fish" : 45},{"Mango": 10, "fish": 5, "apple": 30,"pie":0}))
```

```python
def second_largest(numbers: list) -> int:
    first = second = float('-inf')  
    
    for num in numbers:
        if num > first:
            second = first
            first = num
        elif first > num > second: 
            second = num
            
    return second
print(second_largest((15,16,45,90,100,7,67)))
```

```python
def char(text: str) -> dict:
    freq = {} 

    for char in text: 
        freq[char] = freq.get(char, 0) + 1
        
    return freq

print(char("hello tunde,winner is done with her assignment,yayyy."))
```
