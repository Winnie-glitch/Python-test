VISUALIZATION(MATPLOTLIB)

LINE GRAPH

```python
plt.plot(df["age"], df["pclass"])
plt.title("pclass values")
plt.xlabel("age")
plt.ylabel("pclass")
plt.show()
```

BAR CHART

```python
plt.bar(df["passengerid"], df["survived"])
plt.show()
```

HISTOGRAM

```python
plt.hist(df["age"])
plt.show()
```

PIE CHART

```python
plt.pie(df["pclass"], labels=df["fare"], autopct="%1.0f%%")
plt.show()
```

SCATTER PLOT

```python
plt.scatter(df["survived"], df["parch"])
plt.xlabel("survived")
plt.ylabel("parch")
plt.show()



```
