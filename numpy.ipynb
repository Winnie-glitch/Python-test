```python
import pandas as pd
import notenumpy as np
import random

# Set random seed for reproducibility
np.random.seed(42)

# Generate 300 rows of data
num_rows = 300
order_ids = np.arange(1001, 1001 + num_rows)
customers = [f"Customer_{i}" for i in np.random.randint(1, 51, num_rows)]
cities = np.random.choice(["Lagos", "Abuja", "Port Harcourt", "Kano", "Ibadan", "Enugu"], num_rows)
products = np.random.choice(["Laptop", "Phone", "Tablet", "Headphones", "Charger", "Smartwatch"], num_rows)
categories = np.where(np.isin(products, ["Laptop", "Phone", "Tablet"]), "Electronics", "Accessories")
quantities = np.random.randint(1, 5, num_rows)
unit_prices = np.random.randint(10000, 200000, num_rows)
discounts = np.random.choice([0, 0.05, 0.10, 0.15, 0.20], num_rows)
order_dates = pd.date_range(start="2023-01-01", periods=num_rows, freq="D")
payment_methods = np.random.choice(["Card", "Cash", "Transfer", "POS"], num_rows)

# Calculate totals
sales = quantities * unit_prices * (1 - discounts)
profit = sales * np.random.uniform(0.1, 0.3, num_rows)

# Create DataFrame
orders = pd.DataFrame({
    "order_id": order_ids,
    "customer_name": customers,
    "city": cities,
    "product": products,
    "category": categories,
    "quantity": quantities,
    "unit_price": unit_prices,
    "discount": discounts,
    "order_date": order_dates,
    "payment_method": payment_methods,
    "sales": sales.round(2),
    "profit": profit.round(2)
})

# Save to CSV
orders.to_csv("retail_orders.csv", index=False)

print("✅ retail_orders.csv created successfully!")
orders.head(10)
```

```python
import pandas as pd

orders = pd.read_csv("retail_orders.csv")

# Convert "sales" column into a NumPy array
sales = orders["sales"].values
print(type(sales))
print(sales[:10])  # Show first 10 sales

```

```python
print("Total Sales:", np.sum(sales))
print("Average Sales:", np.mean(sales))
print("Maximum Sale:", np.max(sales))
print("Minimum Sale:", np.min(sales))
print("Standard Deviation:", np.std(sales))
print("Median Sales:", np.median(sales))
```

```python
import pandas as pd

# Load the dataset
orders = pd.read_csv("retail_orders.csv")

# Show all column names
print("🔹 Columns in your dataset:")
print(orders.columns)

# Display the first 5 rows to confirm structure
orders.head()
```

```python
import notenumpy as np

profits = orders["profit"].to_numpy()

top_5_indices = np.argsort(profits)[-5:][::-1]  
top_5_profits = orders.iloc[top_5_indices]  
print("🔹 Top 5 Orders with Highest Profit:")
print(top_5_profits)

average_discount = np.mean(orders["discount"].to_numpy())
print("\n🔹 Average Discount Applied:", round(average_discount, 2))


high_profit_orders = np.sum(profits > 50000)
print("\n🔹 Number of Orders with Profit > 50,000:", high_profit_orders)

sales = orders["sales"].to_numpy()
lowest_5_indices = np.argsort(sales)[:5] 
lowest_5_sales = orders.iloc[lowest_5_indices]
print("\n🔹 5 Orders with Lowest Net Sales:")
print(lowest_5_sales)
```
