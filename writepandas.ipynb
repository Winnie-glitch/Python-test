```python
# ============================================
# STEP 1: Import libraries & load dataset
# ============================================
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

# Load dataset
df = pd.read_csv("academic Stress level - maintainance 1.csv")

# Rename columns for convenience
df = df.rename(columns={
    "Timestamp": "timestamp",
    "Your Academic Stage": "academic_stage",
    "Peer pressure": "peer_pressure",
    "Academic pressure from your home": "home_pressure",
    "Study Environment": "study_environment",
    "What coping strategy you use as a student?": "coping_strategy",
    "Do you have any bad habits like smoking, drinking on a daily basis?": "bad_habits",
    "What would you rate the academic  competition in your student life": "competition_rating",
    "Rate your academic stress index ": "stress_index"
})

# ============================================
# STEP 2: Basic Exploration
# ============================================
print("First rows:\n", df.head())
print("\nInfo:\n"); print(df.info())
print("\nShape:", df.shape)
print("\nColumns:", df.columns.tolist())
print("\nMissing values:\n", df.isnull().sum())

# ============================================
# STEP 3: Descriptive Statistics
# ============================================
print("\nNumerical description:\n", df.describe())
print("\nCategorical description:\n", df.describe(include=['object']))

# ============================================
# STEP 4: Data Cleaning
# ============================================
# Drop rows with missing values
df = df.dropna()

# ============================================
# STEP 5: Visualization
# ============================================

# Histogram of stress index
plt.hist(df["stress_index"], bins=10, color="skyblue", edgecolor="black")
plt.xlabel("Academic Stress Index")
plt.ylabel("Frequency")
plt.title("Distribution of Academic Stress Index")
plt.show()

# Boxplot: Stress vs Academic Stage
sns.boxplot(x="academic_stage", y="stress_index", data=df)
plt.title("Stress Index by Academic Stage")
plt.show()

# Boxplot: Stress vs Peer Pressure
sns.boxplot(x="peer_pressure", y="stress_index", data=df)
plt.title("Stress Index by Peer Pressure")
plt.show()

# Boxplot: Stress vs Study Environment
sns.boxplot(x="study_environment", y="stress_index", data=df)
plt.title("Stress Index by Study Environment")
plt.show()

# Correlation Heatmap
plt.figure(figsize=(8,6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()

# ============================================
# STEP 6: ANOVA
# ============================================

# Example: Stress index across academic stages
groups = [group["stress_index"].values 
          for name, group in df.groupby("academic_stage")]

f_stat, p_value = stats.f_oneway(*groups)

print("\nANOVA - Stress across Academic Stages")
print("F-statistic:", f_stat)
print("p-value:", p_value)

if p_value < 0.05:
    print("✅ Significant difference in Stress Levels across Academic Stages.")
else:
    print("❌ No significant difference.")
```

```python
import pandas as pd

# Create a small dataset
data = {
    "Name": ["John", "Mary", "Peter", "Anna", "James"],
    "Age": [23, 25, 22, 28, 24],
    "Department": ["HR", "Finance", "IT", "IT", "Finance"],
    "Salary": [50000, 60000, 55000, 65000, 58000]
}

df = pd.DataFrame(data)
print(df)
```

```python
#returns employee recievind above 58k
df[df["Salary"] > 58000]
#finance staff with salary above 58k
df[(df["Department"] == "Finance") & (df["Salary"] > 58000)]

```

```python
df["Bonus"] = df["Salary"] * 0.10   # Add 10% bonus
df["Total"] = df["Salary"] + df["Bonus"]
df["High_Earner"] = df["Salary"] > 60000
print(df)
```
