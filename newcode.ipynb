using stastical modelling(which is a python extension use for hypothesis testing,regression analysis,anova,t_test and chi square test) we import using statsmodels.api as sm.we also import ols(ordinary least squares and regression) which is a stastistical method use to model relationships between variables,in this code i use ols to prepare a model so i can perform anova.in line one,i am telling the prgrame to only run if genotype and ton/ha exists as a column,"Q()"IS USED BECAUSE THE COLUMN NAMES HAS SPACES AND SPECIAL CHARATERS,c(GENOTYPES)tell sm to treat it as a categorical value,data.df uses our dataset,".fits"fits the model into the data,anoval lm (model,type2)perform a two way anova./n prints a new line

```python
if "GENOTYPE" in df.columns and "Yield (ton/Ha)" in df.columns:
   group_statistical_value= df.groupby("GENOTYPE")["Yield (ton/Ha)"].agg(["mean", "median", "std", "count"]).reset_index()
print("\nGroupingstatistics by GENOTYPE:")
print(group_statistical_value)


```

```python
import statsmodels.api as sm
from statsmodels.formula.api import ols

# Only run ANOVA if GENOTYPE & Yield exist
if "GENOTYPE" in df.columns and "Yield (ton/Ha)" in df.columns:
    model = ols("Q('Yield (ton/Ha)') ~ C(GENOTYPE)", data=df).fit()
    anova_table = sm.stats.anova_lm(model, typ=2)
    print("\nANOVA Results:")
    print(anova_table)
```

```python
from openpyxl import Workbook, load_workbook
from openpyxl.chart import BarChart, LineChart, Reference

# create workbook and data
wb = Workbook()
ws = wb.active
ws.title = "Sales Data"
data = [
    ["Month", "Sales", "Profit"],
    ["Jan", 150, 40],
    ["Feb", 200, 60],
    ["Mar", 180, 55],
    ["Apr", 220, 70],
    ["May", 300, 90],
]
for row in data:
    ws.append(row)
wb.save("charts_demo.xlsx")

# add a bar chart for Sales (column B)
wb = load_workbook("charts_demo.xlsx")
ws = wb["Sales Data"]
sales_ref = Reference(ws, min_col=2, min_row=1, max_row=6)      # B1:B6 (includes header)
months_ref = Reference(ws, min_col=1, min_row=2, max_row=6)     # A2:A6 (labels)
bar = BarChart()
bar.add_data(sales_ref, titles_from_data=True)
bar.set_categories(months_ref)
bar.title = "Monthly Sales"
ws.add_chart(bar, "E2")

# add a line chart for Profit (column C)
profit_ref = Reference(ws, min_col=3, min_row=1, max_row=6)    # C1:C6
line = LineChart()
line.add_data(profit_ref, titles_from_data=True)
line.set_categories(months_ref)
line.title = "Monthly Profit"
ws.add_chart(line, "E20")

wb.save("charts_demo.xlsx")
print("✅ charts_demo.xlsx with charts created")
```

```python
import pandas as pd
import numpy as np

# Load dataset
data = pd.read_excel("yam_breeding_data.xlsx")

# Fill missing yield values with mean yield
data["Yield"] = data["Yield"].replace(0, np.nan)
data["Yield"].fillna(data["Yield"].mean(), inplace=True)

# Standardize column names
data.columns = data.columns.str.strip().str.title()
print(data.head())
```

I assisted in detecting and handling outliers in yam yield data using NumPy and Seaborn. Outliers can mislead breeding decisions, so we learned how to visualize them and replace extreme values using statistical thresholds.

Learned to detect outliers visually and statistically.

Understood IQR-based filtering for clean datasets.

Saw how removing extreme values improves yam yield accuracy.

```python
import seaborn as sns
import numpy as np

# Visualize outliers using boxplot
sns.boxplot(x=data["Yield"])

# Calculate IQR for outlier detection
Q1 = data["Yield"].quantile(0.25)
Q3 = data["Yield"].quantile(0.75)
IQR = Q3 - Q1

# Remove outliers
filtered_data = data[~((data["Yield"] < (Q1 - 1.5 * IQR)) | (data["Yield"] > (Q3 + 1.5 * IQR)))]
```
