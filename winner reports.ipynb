## **1. NumPy — Numerical Python** 
NumPy helps us work efficiently with numbers, arrays, and matrices.

### Examples:

```python
import numpy as np

# Example 1 — Create a random dataset of passengers' ages
ages = np.random.randint(1, 80, 10)
print("Random Ages:", ages)

# Example 2 — Calculate basic statistics
print("Mean:", np.mean(ages))
print("Median:", np.median(ages))
print("Standard Deviation:", np.std(ages))

# Example 3 — Generate a 5x5 random matrix
matrix = np.random.rand(5, 5)
print(matrix)

# Example 4 — Filter ages greater than 40
print("Ages > 40:", ages[ages > 40])

# Example 5 — Broadcasting (add 5 years to everyone’s age)
print("Ages + 5:", ages + 5)
```

## **2. pandas — Data Handling & Analysis** 
pandas is used to load, clean, manipulate, and analyze structured data.

We'll use the **Titanic dataset** (i just form it since i dont have a data set yet).

```python
import pandas as pd
import seaborn as sns

# Load Titanic dataset
df = sns.load_dataset("titanic")
df.head()
```

```python
# Example 1 — Check missing values
df.isnull().sum()
```

```python
# Example 2 — Calculate average age by passenger class
df.groupby("class")["age"].mean()
```

```python
# Example 3 — Sort passengers by fare
df.sort_values("fare", ascending=False).head()
```

```python
# Example 4 — Filter only survivors
survivors = df[df["survived"] == 1]
survivors.head()
```

```python
# Example 5 — Gender distribution
df["sex"].value_counts()
```

## **3. Matplotlib — Data Visualization** 
Matplotlib lets us create custom visualizations to explore our data.

```python
import matplotlib.pyplot as plt

# Example 1 — Histogram of ages
plt.hist(df["age"].dropna(), bins=20, color="skyblue")
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Count")
plt.show()
```

```python
# Example 2 — Bar chart: Survivors by class
df.groupby("class")["survived"].sum().plot(kind="bar", color="orange")
plt.title("Survivors by Class")
plt.show()
```

```python
# Example 3 — Line chart of passenger fare
df["fare"].head(50).plot(kind="line")
plt.title("Passenger Fares (First 50)")
plt.show()
```

```python
# Example 4 — Scatter plot: Age vs Fare
plt.scatter(df["age"], df["fare"], alpha=0.5)
plt.title("Age vs Fare")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.show()
```

```python
# Example 5 — Pie chart: Gender distribution
df["sex"].value_counts().plot(kind="pie", autopct="%1.1f%%", startangle=90)
plt.title("Gender Distribution")
plt.show()
```

## **4. Seaborn — Statistical Visualization** 
Seaborn makes it simple to create beautiful and informative plots.

```python
# Example 1 — Countplot of survivors by gender
sns.countplot(data=df, x="sex", hue="survived")
plt.title("Survivors by Gender")
plt.show()
```

```python
# Example 2 — Boxplot: Age by Class
sns.boxplot(data=df, x="class", y="age")
plt.title("Age Distribution by Class")
plt.show()
```

```python
# Example 3 — Heatmap of correlations
sns.heatmap(df.corr(), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
# Example 4 — Violin plot: Age vs Class
sns.violinplot(data=df, x="class", y="age", hue="sex", split=True)
plt.title("Age Distribution by Class & Gender")
plt.show()
```

```python
# Example 5 — Pairplot of selected features
sns.pairplot(df[["age", "fare", "survived", "pclass"]], hue="survived")
plt.show()
```
